# MRI — from intuition to engineering

**Start here.** This folder is a self-contained course on how magnetic resonance imaging (MRI) works. It covers the physics, the mathematics, and the hardware and engineering that make a scanner work.

It is an adaptation and extension of Peder Larson's open textbook [*Principles of MRI*](https://larsonlab.github.io/MRI-education-resources) (MIT License). The original is excellent but assumes a physics or engineering background: many symbols are used without definition, and steps in derivations are skipped. This version:

* **defines every symbol** in each notebook (a symbol table at the top of every chapter, plus a [master notation reference](reference/Notation_and_Symbols.ipynb)),
* follows each equation with an **"in words"** explanation and its units,
* adds missing derivations as **optional, collapsible blocks**,
* adds **hardware and engineering chapters** (magnet, gradients, RF chain, safety),
* ports all of the book's Octave/MATLAB code to **runnable Python**, and
* adds optional **refresher notebooks** for the math and physics background.

## How the course is organised: two passes

The course deliberately does **not** climb slowly from "easy" to "hard" in one long line. It goes over the whole subject **twice**:

| | Pass 1 — the big picture | Pass 2 — the deep dive |
|---|---|---|
| Goal | Have the *whole map* of MRI in your head: physics, scanner, image formation, contrast, speed, artifacts, safety | Revisit every topic rigorously: derivations, simulations, hardware details, engineering trade-offs |
| Math | Minimal; every symbol still defined | Full; each step explained, with links to refreshers |
| Length | 5 notebooks, ~30–45 min each | 34 chapters in 10 parts (A–J) |
| Folder | `pass1_big_picture/` | `pass2_deep_dive/` |

The idea: in Pass 2 you always know *why* a detail matters, because Pass 1 has already shown you where it fits.

**Refreshers** (`refreshers/`) are optional side trips. A chapter links to one exactly where you need it. **Reference** notebooks (`reference/`) are for looking things up: notation, glossary, key equations, practice questions.

## Conventions used everywhere

You only need to know four things up front. The [notation reference](reference/Notation_and_Symbols.ipynb) has everything else.

1. **SI units in the code**: seconds, tesla, hertz, metres. The text often says "ms", "mT" or "MHz" for readability, and the code converts explicitly.
2. **Two gyromagnetic ratios**: $\bar\gamma = 42.58$ MHz/T (hertz per tesla) and $\gamma = 2\pi\bar\gamma \approx 2.675\times10^8$ rad s⁻¹ T⁻¹. Mixing them up is the most common source of stray factors of $2\pi$.
3. **Precession is clockwise** (left-handed): a transverse magnetization written as a complex number, $M_{xy}=M_x+iM_y$, evolves as $M_{xy}(t)=M_{xy}(0)\,e^{-i\omega t}$.
4. **Fourier transform convention**: $F(k)=\int f(x)\,e^{-i2\pi kx}\,dx$, with $k$ in cycles per metre.

Every notebook imports the shared helper module [`mri_tools.py`](mri_tools.py), which contains the Bloch simulator, rotations, centred FFTs, phantoms, signal equations and coil-field calculations. Reading its top-of-file docstring is a good 5-minute investment. The cell below checks that your environment works:

````python
import sys, os
sys.path.insert(0, os.path.abspath("."))   # mri_tools.py lives in this folder
import numpy as np
import matplotlib.pyplot as plt
import mri_tools as mt
mt.setup_plots()
%matplotlib inline

print(f"Python {sys.version.split()[0]}, numpy {np.__version__}")
for B0 in (0.55, 1.5, 3.0, 7.0):
    print(f"B0 = {B0:4.2f} T  ->  1H Larmor frequency f0 = {mt.larmor_frequency(B0)/1e6:6.2f} MHz")

# A first taste: magnetization tipped by 90 degrees, then precessing and relaxing
# (rotating frame, 20 Hz off-resonance, T1 = 1 s, T2 = 100 ms)
t, M = mt.bloch_simulate(n_steps=400, dt=1e-3, M_init=[0, 1, 0], T1=1.0, T2=0.1, df=20)
fig, ax = plt.subplots(figsize=(8, 3.5))
mt.plot_magnetization(t, M, ax=ax, components="xyza",
                      title="After a 90° pulse: precession (Mx, My), T2 decay of |Mxy|, T1 recovery of Mz")
plt.show()
````

In the plot, $M_x$ and $M_y$ oscillate (precession at 20 Hz relative to the rotating frame) inside a decaying envelope $|M_{xy}|$ ($T_2$ = 100 ms), while $M_z$ slowly grows back toward 1 ($T_1$ = 1 s). By the end of Pass 1 every word in that sentence will be familiar, and by the end of Pass 2 you will be able to derive it.
